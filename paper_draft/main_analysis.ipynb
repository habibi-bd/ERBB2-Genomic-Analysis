```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from rdkit import Chem
from rdkit.Chem.MolStandardize import rdMolStandardize

def standardize_smiles_complete(smiles_string):
    try:
        if pd.isna(smiles_string):
            return None
            
        # Task 1: Molecule Parsing & Initial Sanitization
        mol = Chem.MolFromSmiles(str(smiles_string))
        if mol is None:
            return None
        
        # Task 2: Salt Stripping through Largest-Fragment Selection
        remover = rdMolStandardize.LargestFragmentChooser()
        mol = remover.choose(mol)
        
        # Task 3: Molecule Normalization (bonds, tautomers, functional groups)
        normalizer = rdMolStandardize.Normalizer()
        mol = normalizer.normalize(mol)
        
        # Task 4: Neutralization (removes formal charges from atoms safely)
        uncharger = rdMolStandardize.Uncharger()
        mol = uncharger.uncharge(mol)
        
        # Task 5: Final Sanitization (validates aromaticity, valence, etc.)
        Chem.SanitizeMol(mol)
        
        # Output: Convert back to standard Canonical SMILES string
        return Chem.MolToSmiles(mol, canonical=True)
        
    except Exception as e:
        return None
# # Example Usage:
# raw_smiles = "[Na+].CC(=O)[O-].CCN(CC)CC" # A mixture with salts and charges
# clean_smiles = standardize_smiles_complete(raw_smiles)
# print(f"Before: {raw_smiles}")
# print(f"After : {clean_smiles}")
```

```python
# Apply the standardization function to every row in the SMILES column
df_cleaned['Standard_SMILES'] = df_cleaned['SMILES'].apply(standardize_smiles_complete)
print(df_cleaned['Standard_SMILES'].isna().sum())
# Drop any rows where RDKit failed to parse the SMILES (returned None)
df_cleaned = df_cleaned.dropna(subset=['Standard_SMILES'])
```

```python

import pandas as pd
from rdkit import Chem
from mordred import Calculator, descriptors
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
import numpy as np

# 2. Initialize Mordred descriptor calculator (1D/2D descriptors only)
calc = Calculator(descriptors, ignore_3D=True)

# 3. Generate descriptors for each standardized SMILES
descriptor_data = []
for smi in df_cleaned['Standard_SMILES']:
    try:
        mol = Chem.MolFromSmiles(smi)
        if mol is not None:
            desc_values = calc(mol)
            descriptor_data.append(desc_values)
        else:
            descriptor_data.append(None)
    except:
        descriptor_data.append(None)

desc_df = pd.DataFrame(descriptor_data)

# 4. Combine descriptors with target column
final_df = pd.concat([df_cleaned[['VERIFY']], desc_df], axis=1)
```

```python
import numpy as np
import pandas as pd  # Ensuring pandas is imported as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

# 5. Remove highly correlated descriptors (Pearson |r| > 0.90)
corr_matrix = final_df.drop(columns=['VERIFY']).corr().abs()

# FIX: Use np directly instead of pd.np
upper = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)
to_drop = [column for column in upper.columns if any(upper[column] > 0.90)]
filtered_df = final_df.drop(columns=to_drop)

# 6. Train/test split
X = filtered_df.drop(columns=['VERIFY'])
y = filtered_df['VERIFY'].astype(int)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# 7. Scale features (fit only on training set to avoid leakage)
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Final descriptor count:", X_train.shape[1])
print("Training set size:", X_train.shape)
print("Test set size:", X_test.shape)
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import (
    roc_auc_score, accuracy_score, recall_score, 
    precision_score, f1_score, matthews_corrcoef, r2_score, confusion_matrix
)
from sklearn.model_selection import StratifiedKFold
from sklearn.base import clone

# Linear Models
from sklearn.linear_model import LogisticRegression, RidgeClassifier, SGDClassifier
from sklearn.svm import LinearSVC

# Probabilistic Models
from sklearn.naive_bayes import GaussianNB
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis, QuadraticDiscriminantAnalysis

# Instance-based
from sklearn.neighbors import KNeighborsClassifier

# Tree & Ensemble Models
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    ExtraTreesClassifier, RandomForestClassifier, BaggingClassifier,
    GradientBoostingClassifier, HistGradientBoostingClassifier, AdaBoostClassifier
)

# Advanced Boosting & Neural Networks
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.neural_network import MLPClassifier
```

```python
def calculate_metrics(y_true, y_pred, y_prob):
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    
    metrics = {
        "ROC AUC": roc_auc_score(y_true, y_prob) if y_prob is not None else np.nan,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Sensitivity": recall_score(y_true, y_pred),  # True Positive Rate
        "Specificity": tn / (tn + fp) if (tn + fp) > 0 else 0,  # True Negative Rate
        "Precision": precision_score(y_true, y_pred, zero_division=0),
        "F1-Score": f1_score(y_true, y_pred, zero_division=0),
        "MCC": matthews_corrcoef(y_true, y_pred),
        "R2 (Prob)": r2_score(y_true, y_prob) if y_prob is not None else np.nan
    }
    return metrics
```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
```

```python
results = []
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

for name, model in models.items():
    print(f"Training and evaluating: {name}...")
    
    cv_metrics = []
    
    for train_idx, val_idx in cv.split(X_train_scaled, y_train):
        X_cv_train, X_cv_val = X_train_scaled[train_idx], X_train_scaled[val_idx]
        y_cv_train, y_cv_val = y_train.iloc[train_idx], y_train.iloc[val_idx]
        
        # Create a pipeline that first imputes missing values, then runs the model
        # 'median' is usually safer than 'mean' if your data has outliers
        pipeline = Pipeline([
            ('imputer', SimpleImputer(strategy='median')),
            ('classifier', clone(model))
        ])
        
        pipeline.fit(X_cv_train, y_cv_train)
        y_cv_pred = pipeline.predict(X_cv_val)
        
        # Extract the classifier step to check for probability methods safely
        clf = pipeline.named_steps['classifier']
        
        if hasattr(clf, "predict_proba"):
            y_cv_prob = pipeline.predict_proba(X_cv_val)[:, 1]
        elif hasattr(clf, "decision_function"):
            y_cv_prob = pipeline.decision_function(X_cv_val)
            y_cv_prob = (y_cv_prob - y_cv_prob.min()) / (y_cv_prob.max() - y_cv_prob.min() + 1e-9)
        else:
            y_cv_prob = y_cv_pred.astype(float)
            
        cv_metrics.append(calculate_metrics(y_cv_val, y_cv_pred, y_cv_prob))
        
    # Average the metrics across all 5 folds
    avg_cv = pd.DataFrame(cv_metrics).mean().to_dict()
    
    # --- Full Train & Test Set Evaluation ---
    full_pipeline = Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('classifier', clone(model))
    ])
    
    full_pipeline.fit(X_train_scaled, y_train)
    y_test_pred = full_pipeline.predict(X_test_scaled)
    
    clf_full = full_pipeline.named_steps['classifier']
    if hasattr(clf_full, "predict_proba"):
        y_test_prob = full_pipeline.predict_proba(X_test_scaled)[:, 1]
    elif hasattr(clf_full, "decision_function"):
        y_test_prob = full_pipeline.decision_function(X_test_scaled)
        y_test_prob = (y_test_prob - y_test_prob.min()) / (y_test_prob.max() - y_test_prob.min() + 1e-9)
    else:
        y_test_prob = y_test_pred.astype(float)
        
    test_metrics = calculate_metrics(y_test, y_test_pred, y_test_prob)
    
    # Format row for final presentation table
    row = {"Model": name}
    for metric in avg_cv.keys():
        row[f"CV_{metric}"] = avg_cv[metric]
        row[f"Test_{metric}"] = test_metrics[metric]
        
    results.append(row)

results_df = pd.DataFrame(results)
```

```python
def calculate_metrics(y_true, y_pred, y_prob):
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    
    metrics = {
        "AUC": roc_auc_score(y_true, y_prob) if y_prob is not None else np.nan,
        "Acc": accuracy_score(y_true, y_pred),
        "Sens": recall_score(y_true, y_pred),
        "Spec": tn / (tn + fp) if (tn + fp) > 0 else 0,
        "Prec": precision_score(y_true, y_pred, zero_division=0),
        "F1": f1_score(y_true, y_pred, zero_division=0),
        "MCC": matthews_corrcoef(y_true, y_pred)
    }
    return metrics
```

```python
results = []
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

for name, model in models.items():
    print(f"Processing: {name}...")
    
    try:
        cv_auc_scores = []
        
        # --- 5-Fold Stratified Cross-Validation ---
        for train_idx, val_idx in cv.split(X_train_scaled, y_train):
            X_cv_train, X_cv_val = X_train_scaled[train_idx], X_train_scaled[val_idx]
            y_cv_train, y_cv_val = y_train.iloc[train_idx], y_train.iloc[val_idx]
            
            pipeline = Pipeline([
                ('imputer', SimpleImputer(strategy='median')),
                ('classifier', clone(model))
            ])
            
            pipeline.fit(X_cv_train, y_cv_train)
            
            # Extract probability/decision weights safely for CV AUC
            clf = pipeline.named_steps['classifier']
            if hasattr(clf, "predict_proba"):
                y_cv_prob = pipeline.predict_proba(X_cv_val)[:, 1]
            elif hasattr(clf, "decision_function"):
                y_cv_prob = pipeline.decision_function(X_cv_val)
                y_cv_prob = (y_cv_prob - y_cv_prob.min()) / (y_cv_prob.max() - y_cv_prob.min() + 1e-9)
            else:
                y_cv_prob = pipeline.predict(X_cv_val).astype(float)
                
            cv_auc_scores.append(roc_auc_score(y_cv_val, y_cv_prob))
            
        # Calculate CV AUC Mean and SD
        cv_auc_mean = np.mean(cv_auc_scores)
        cv_auc_sd = np.std(cv_auc_scores)
        
        # --- Full Train & Test Set Evaluation ---
        full_pipeline = Pipeline([
            ('imputer', SimpleImputer(strategy='median')),
            ('classifier', clone(model))
        ])
        full_pipeline.fit(X_train_scaled, y_train)
        
        y_test_pred = full_pipeline.predict(X_test_scaled)
        
        clf_full = full_pipeline.named_steps['classifier']
        if hasattr(clf_full, "predict_proba"):
            y_test_prob = full_pipeline.predict_proba(X_test_scaled)[:, 1]
        elif hasattr(clf_full, "decision_function"):
            y_test_prob = full_pipeline.decision_function(X_test_scaled)
            y_test_prob = (y_test_prob - y_test_prob.min()) / (y_test_prob.max() - y_test_prob.min() + 1e-9)
        else:
            y_test_prob = y_test_pred.astype(float)
            
        test_metrics = calculate_metrics(y_test, y_test_pred, y_test_prob)
        
        # Build the exact table structure matching your image layout
        row = {
            "Model": name,
            "CV AUC ± SD": f"{cv_auc_mean:.3f} ± {cv_auc_sd:.3f}",
            "Test AUC": round(test_metrics["AUC"], 3),
            "Acc": round(test_metrics["Acc"], 3),
            "Sens": round(test_metrics["Sens"], 3),
            "Spec": round(test_metrics["Spec"], 3),
            "Prec": round(test_metrics["Prec"], 3),
            "F1": round(test_metrics["F1"], 3),
            "MCC": round(test_metrics["MCC"], 3)
        }
        results.append(row)
        
    except Exception as e:
        print(f"⚠️ Skipping {name}: {e}")
        continue

# Package into clean final table format
final_metrics_df = pd.DataFrame(results)
```
