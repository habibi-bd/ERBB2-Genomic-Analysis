```python
import pandas as pd

def prepare_and_save_training_set():
    print("--- STEP 1: LOADING RAW CHEMBL DATA ---")
    # Load your raw ChEMBL bioactivity dataset
    df_raw = pd.read_csv("HER2_ChEMBL_bioactivity_data.csv")
    
    print(f"Initial raw dataset shape: {df_raw.shape}")
    
    # Clean up null values or missing critical elements
    df_clean = df_raw.dropna(subset=['canonical_smiles', 'activity_class']).copy()
    
    # Map text activity classes ('active' / 'inactive') to clean binary integers (1 / 0)
    # This aligns directly with the paper's binary classification strategy
    df_clean['binary_outcome'] = df_clean['activity_class'].apply(lambda x: 1 if str(x).lower() == 'active' else 0)
    
    # Keep only the essential columns to make the training set clean and light
    prepared_df = df_clean[['molecule_chembl_id', 'canonical_smiles', 'activity_class', 'binary_outcome']]
    
    # Drop any duplicate chemical entries to prevent data leakage during model training
    prepared_df = prepared_df.drop_duplicates(subset=['canonical_smiles'])
    
    # Save the prepared dataset to a new CSV file
    output_filename = "data/HER2_ChEMBL_prepared_training_set.csv"
    prepared_df.to_csv(output_filename, index=False)
    
    print(f"\n🎉 Success! Prepared training dataset saved to '{output_filename}'")
    print(f"Final Count: {prepared_df.shape[0]} unique compounds.")
    print(f" -> Actives (1): {sum(prepared_df['binary_outcome'] == 1)}")
    print(f" -> Inactives (0): {sum(prepared_df['binary_outcome'] == 0)}")

# Run the preparation script
prepare_and_save_training_set()
```
