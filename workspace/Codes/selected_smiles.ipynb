```python
import pandas as pd
import requests

# 1. Your list of Top 10 ranked Molecule IDs
top_10_ids = [
    130275856, 130276350, 130275835, 130276911, 130276952,
    130276411, 130276915, 130277015, 155756034, 130276381
]

# Convert list to a comma-separated string for a fast batch API request
id_string = ",".join(map(str, top_10_ids))
url = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/cid/{id_string}/property/Title,IUPACName/JSON"

# 2. Fetch names from PubChem
print("Fetching molecule names from PubChem...\n")
name_map = {}

try:
    response = requests.get(url, timeout=10)
    if response.status_code == 200:
        properties_list = response.json()['PropertyTable']['Properties']
        for prop in properties_list:
            cid = prop['CID']
            # Prioritize a short common name (Title), fallback to full IUPACName
            name = prop.get('Title') or prop.get('IUPACName') or "Unknown_Compound"
            name_map[cid] = name.replace(" ", "_") # Keeps format clean
except Exception as e:
    print(f"Error fetching data: {e}")

# 3. Print out the final paired list in order of your ranking
for cid in top_10_ids:
    molecule_name = name_map.get(cid, "Name_Not_Found")
    print(f"{cid} {molecule_name}")
```

```python
import pandas as pd
import requests
import time

# 1. Sample of your dataframe
# df = pd.DataFrame({'ConnectivitySMILES': [...]})

def smiles_to_name(smiles):
    """Queries PubChem API to get the preferred IUPAC/Common name for a SMILES string."""
    if pd.isna(smiles) or not isinstance(smiles, str):
        return None
    
    # URL encode the SMILES string to handle special characters like # or +
    url = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/smiles/{requests.utils.quote(smiles)}/property/IUPACName,Title/JSON"
    
    try:
        response = requests.get(url, timeout=5)
        if response.status_code == 200:
            data = response.json()
            # Try to get the short 'Title' (often common name), fallback to full IUPAC name
            properties = data['PropertyTable']['Properties'][0]
            return properties.get('Title') or properties.get('IUPACName')
    except Exception as e:
        pass # Handle timeouts or network blips gracefully
    
    return None

# 2. Apply the function to your column
# Note: If your dataframe is very large, consider adding a tiny sleep delay to respect API limits
print("Fetching molecule names from PubChem...")
df['MoleculeName'] = df['ConnectivitySMILES'].apply(smiles_to_name)

print(df[['ConnectivitySMILES', 'MoleculeName']].head())
```

```python
import pandas as pd
import requests

# 1. Assuming your dataframe is already loaded as 'df'
# df = pd.DataFrame({'ConnectivitySMILES': [...]})

# Create a cache to avoid duplicate API requests for identical SMILES
smiles_cache = {}

def get_molecule_name(smiles):
    if pd.isna(smiles) or not isinstance(smiles, str) or '...' in smiles:
        return "Unknown_Structure"
    
    # Return cached result if we've already fetched this molecule
    if smiles in smiles_cache:
        return smiles_cache[smiles]
    
    url = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/smiles/{requests.utils.quote(smiles)}/property/Title,IUPACName/JSON"
    try:
        response = requests.get(url, timeout=5)
        if response.status_code == 200:
            properties = response.json()['PropertyTable']['Properties'][0]
            # Prioritize 'Title' (Common name) over long IUPAC string
            name = properties.get('Title') or properties.get('IUPACName') or "Unknown"
            # Remove spaces from the name if you want a strict "SMILES Name" single-space formatting
            name = name.replace(" ", "_") 
            smiles_cache[smiles] = name
            return name
    except Exception:
        pass
    
    return "Unknown"

for smiles in df['ConnectivitySMILES']:
    # Clean the string to ensure no trailing whitespaces impact the API
    clean_smiles = str(smiles).strip()
    molecule_name = get_molecule_name(clean_smiles)
    
    print(f"{clean_smiles} {molecule_name}")
```
