```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Data setup
models = ['SVM-Morgan3', 'SVM-MACCS', 'RF-MACCS', 'RF-Morgan3']
baseline_acc = [0.60, 0.85, 0.84, 0.83]
baseline_auc = [0.58, 0.91, 0.91, 0.91]

tuned_acc = [0.87, 0.85, 0.85, 0.86]  # Matching order: SVM-Morgan3, SVM-MACCS, RF-MACCS, RF-Morgan3
tuned_auc = [0.93, 0.92, 0.92, 0.93]

x = np.arange(len(models))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 5), dpi=300)
rects1 = ax.bar(x - width/2, baseline_auc, width, label='Baseline AUC', color='#99b898')
rects2 = ax.bar(x + width/2, tuned_auc, width, label='Fine-Tuned AUC', color='#e84a5f')

ax.set_ylabel('AUC-ROC Score', fontsize=12, fontweight='bold')
ax.set_title('Performance Comparison: Baseline vs. Fine-Tuned Models (AUC)', fontsize=14, fontweight='bold', pad=15)
ax.set_xticks(x)
ax.set_xticklabels(models, fontsize=11)
ax.set_ylim(0.4, 1.0)
ax.legend(frameon=True, facecolor='white', edgecolor='none')
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.savefig('baseline_vs_tuned_auc.png')
print("Graph generated successfully.")
```
