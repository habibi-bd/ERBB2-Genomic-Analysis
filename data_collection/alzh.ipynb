```python
# 1. Calculate the percentage of null values for each column
null_pct = alzh.isnull().mean() * 100

# 2. Filter for columns that have nulls, but less than 10%
# (We check > 0 so we don't accidentally display columns with 0 missing values)
low_null_cols = null_pct[(null_pct > 0) & (null_pct < 10)]

# 3. Display the exact counts for those specific columns
print(alzh[low_null_cols.index].isnull().sum())
```

```python
import pandas as pd

# 1. Load the data safely, skipping any malformed rows that cause ParserError
# (We use on_bad_lines='skip' to automatically bypass lines with formatting issues)
alzh = pd.read_csv('alzh.csv', on_bad_lines='skip')

# 2. List of columns that are safe to drop for biological/chemical analysis
columns_to_drop = [
    'Journal', 
    'Volume', 
    'First Page', 
    'Last Page', 
    'Variant Sequence Accession', 
    'Variant Sequence Mutation'
]

# 3. Drop the columns if they exist in your dataframe
# (errors='ignore' ensures the code won't crash if a column name is slightly different)
alzh_cleaned = alzh.drop(columns=columns_to_drop, errors='ignore')

# 4. Save the cleaned data to a new file so you don't overwrite your original data
alzh_cleaned.to_csv('alzh_cleaned.csv', index=False)

# Inspect the remaining columns
print("Remaining features for analysis:")
print(alzh_cleaned.columns.tolist())
print(f"\nNew data shape: {alzh_cleaned.shape}")
```

```python
import pandas as pd

# Let's force pandas to properly parse it by reading the original file directly 
# or fixing the cleaned one. We will read 'alzh.csv' properly from the start.
df = pd.read_csv('alzh.csv', sep=';', on_bad_lines='skip')

# Strip out any hidden whitespaces from column names just in case
df.columns = df.columns.str.strip()

# Now drop the unneeded columns safely
columns_to_drop = [
    'Journal', 'Volume', 'First Page', 'Last Page', 
    'Variant Sequence Accession', 'Variant Sequence Mutation',
    'PubMed ID', 'DOI', 'Year'
]
alzh_cleaned = df.drop(columns=columns_to_drop, errors='ignore')

# Save it using a standard comma separator this time!
alzh_cleaned.to_csv('alzh_fully_cleaned.csv', index=False)

# Verify by showing the first 2 rows and the actual column list
print("Columns successfully parsed:")
print(alzh_cleaned.columns.tolist())
print("\nFirst row preview:")
print(alzh_cleaned.head(1))
```

```python
import pandas as pd

# Load our beautifully cleaned data
df = pd.read_csv('alzh_fully_cleaned.csv')

print("=== 1. Top Target Proteins / Genes ===")
# This tells us which specific disease targets are most heavily studied in your file
print(df['Target ChEMBL ID'].value_counts().head(10))

print("\n=== 2. Top Organisms in Dataset ===")
# We want 'Homo sapiens' for human disease analysis
print(df['Organism'].value_counts().head(5))

print("\n=== 3. Assay Types ===")
# 'B' = Binding affinity (how well it sticks), 'F' = Functional (what it does to the cell)
print(df['Assay Type'].value_counts())
```

```python
# 1. Filter for human targets
human_data = df[df['Organism'] == 'Homo sapiens']

# 2. See the most tested targets in your dataset
print("Top Human Targets in Dataset:")
print(human_data['Target ChEMBL ID'].value_counts().head(10))
```
